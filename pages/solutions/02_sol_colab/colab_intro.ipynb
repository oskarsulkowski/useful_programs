## Opis obliczeń

Rozważamy sumę n wyrazów ciągu geometrycznego:

S_n = a + ar + ar^2 + ... + ar^(n-1)

gdzie:
- a — pierwszy wyraz,
- r — iloraz,
- n — liczba wyrazów.

Wzór na sumę:
$$
S_n = a \cdot \frac{1-r^n}{1-r}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation

# Parametry wahadła
g = 9.81
L = 1.0
theta0 = np.pi / 4
dt = 0.02
t = np.arange(0, 10, dt)

theta = theta0 * np.cos(np.sqrt(g / L) * t)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.axis('off')

rod, = ax.plot([], [], 'o-', lw=2, color='blue')
mass, = ax.plot([], [], 'o', markersize=10, color='red')

def init():
    rod.set_data([], [])
    mass.set_data([], [])
    return rod, mass

def animate(i):
    x = L * np.sin(theta[i])
    y = -L * np.cos(theta[i])
    rod.set_data([0, x], [0, y])
    mass.set_data([x], [y])
    return rod, mass

anim = animation.FuncAnimation(
    fig, animate, init_func=init, frames=len(t), interval=20, blit=True
)

anim.save("video.mp4", writer="ffmpeg", fps=30)
print("Zapisano video.mp4")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Modyfikacja krzywej parametrycznej
t = np.linspace(0, 2 * np.pi, 500)
x = np.sin(t) + 0.5 * np.cos(3 * t)
y = np.cos(t) + 0.5 * np.sin(2 * t)
z = 0.5 * t / (2 * np.pi)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="Modified Parametric Curve")
ax.set_title("Zmieniona krzywa parametryczna", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()
plt.show()

fig.savefig("parametric_curve.png")
print("Zapisano parametric_curve.png")
```
