
### Opis obliczeń

Rozważmy ciąg geometryczny:
$$a_n = a_1\, r^{\,n-1}.$$

Suma pierwszych n wyrazów ciągu geometrycznego (dla \(r \neq 1\)):
$$
S_n \;=\; \sum_{k=1}^{n} a_1 r^{\,k-1}
\;=\; a_1 \frac{1 - r^{n}}{1 - r}, \qquad r \ne 1.
$$

Dowód (krótko): pomnóż obie strony sumy przez \(r\) i odejmij:
\[
(1-r)S_n = a_1(1-r^n) \quad\Rightarrow\quad S_n = a_1\frac{1-r^n}{1-r}.
\]

Przypadek szczególny: jeśli \(r=1\), to \(S_n = n\,a_1\).

Przykład:
- \(a_1 = 2\)  
- \(r = 3\)  
- \(n = 5\)

Wtedy:

$$
S_5 \;=\; 2 \cdot \frac{1-3^{5}}{1-3}
\;=\; 2 \cdot \frac{1-243}{1-3}
\;=\; 2 \cdot \frac{-242}{-2}
\;=\; 2 \cdot 121
\;=\; 242.
$$

```python
#utworzenie filmu wideo.mp4 z animacją wahadła
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Parametry wahadła
L = 1.0      # długość wahadła [m]
g = 9.81     # przyspieszenie ziemskie [m/s^2]
theta0 = np.deg2rad(35)  # początkowy kąt
t_end = 8.0  # czas symulacji [s]
fps = 30
n_frames = int(t_end * fps)

# Czas
t = np.linspace(0, t_end, n_frames)

# Ruch wahadła: theta(t) = theta0 * cos(sqrt(g/L) * t)
theta = theta0 * np.cos(np.sqrt(g / L) * t)

# Pozycja końca wahadła
x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 0.5)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
ax.set_title("Animacja wahadła")
ax.axhline(0, color="black", lw=1)

pivot, = ax.plot([0], [0], marker='o', markersize=10, color='black')
rod, = ax.plot([], [], color='black', lw=2)
bob, = ax.plot([], [], marker='o', markersize=14, color='tab:red')

def update(frame):
    rod.set_data([0, x[frame]], [0, y[frame]])
    bob.set_data([x[frame]], [y[frame]])
    return rod, bob

anim = FuncAnimation(fig, update, frames=np.arange(n_frames), interval=1000/fps, blit=True)
anim.save("video.mp4", writer="ffmpeg", fps=fps)
plt.close(fig)
print("Zapisano wideo do: video.mp4")
```

```python
#zmodyfikowana krzywa parametryczna, wyświetlenie i zapis do pliku
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # potrzebne do rysowania 3D

#Zmieniona krzywa parametryczna: ślimak / toroidalna
t = np.linspace(0, 4 * np.pi, 1000)

#Modyfikacja oryginalnej krzywej:
x = (2 + np.cos(3 * t)) * np.cos(t)
y = (2 + np.cos(3 * t)) * np.sin(t)
z = np.sin(5 * t)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, color='purple', linewidth=2, label="Zmodyfikowany Parametric Curve")
ax.set_title("Zmodyfikowana krzywa parametryczna", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()
plt.tight_layout()
plt.savefig("Zmodyfikowany_parametric_curve.png", dpi=200)
plt.show()
print("Zapisano krzywą do: Zmodyfikowany_parametric_curve.png")
```
